```python
df['avgrage']=df['total']/3
df
```

```python
# 5. Introduce a missing value (NaN) in the "Science" column and then:
#     (a) Identify missing values using isnull()
df.loc[2,'science']=None
df
```

```python
df['science'].fillna(df['science'].mean(),inplace=True)
df
```

```python
# 6. Filter and display all students whose "Average" score is greater than 75.
df[df['avgrage']>75]
```

```python
# 7. Sort the DataFrame by "Total_Marks" in descending order and show
#the top 3 performers.
df3=df.sort_values(by=['total'],ascending=False)
df3
```
