```python
# 12. Create a function that accepts text input and returns a list of nouns and verbs 
# separately using POS tagging.


def demo(t):
    w = nltk.word_tokenize(t)
    p = nltk.pos_tag(w)

    n = []
    v = []

    for x, y in p:
        if y.startswith('NN'):
            n.append(x)
        elif y.startswith('VB'):
            v.append(x)

    return n, v



s = "rahul is learning Python and make projects"
a, b = demo(s)

print("Nouns:", a)
print("Verbs:", b)
```

```python
# 13. Count the number of adjectives and adverbs present in a given text.

def demo(t):
    w = nltk.word_tokenize(t)
    p = nltk.pos_tag(w)

    a = 0   
    d = 0   

    for x, y in p:
        if y.startswith('JJ'):   # Adjectives
            a = a + 1
        elif y.startswith('RB'): # Adverbs
            d = d + 1

    return a, d


s = "rahul is a very intelligent and extremely hardworking student"
a, d = demo(s)

print("Adjectives:", a)
print("Adverbs:", d)
```

```python
import nltk
from nltk.corpus import movie_reviews, stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
from nltk.probability import FreqDist
from nltk.classify import NaiveBayesClassifier

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('movie_reviews')
nltk.download('wordnet')

docs = [(list(movie_reviews.words(fileid)), category) 
        for category in movie_reviews.categories() 
        for fileid in movie_reviews.fileids(category)]

all_words = [w.lower() for w in movie_reviews.words()]
freq = FreqDist(all_words)
fw = list(freq.keys())[:2000]

def extract_features(words):
    words_set = set(words)
    features = {}
    for x in fw:
        features[x] = (x in words_set)
    return features

feature_sets = [(extract_features(w), c) for (w, c) in docs]
train_set = feature_sets[:1900]
clf = NaiveBayesClassifier.train(train_set)

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    tokens = word_tokenize(text.lower())
    filtered = [w for w in tokens if w.isalpha() and w not in stop_words]
    lemmatized = [lemmatizer.lemmatize(w) for w in filtered]
    return lemmatized

sentences = [
    "The film was boring and way too long",
    "What a brilliant performance by the lead actor",
   
]

for s in sentences:
    words = preprocess(s)
    features = extract_features(words)
    sentiment = clf.classify(features)
    print(f"Sentence: {s}")
    print(f"Predicted sentiment: {sentiment}\n")
```

```python
# 24. Prepare feature sets using the movie_reviews dataset and split
#them into training and testing data.


import random



docs = []

for c in movie_reviews.categories():   
    for f in movie_reviews.fileids(c):
        docs.append((movie_reviews.words(f), c))


random.shuffle(docs)


all_words = []

for w in movie_reviews.words():
    all_words.append(w.lower())


fd = nltk.FreqDist(all_words)

fw = list(fd.keys())[:2000]


def features(doc):
    d = {}
    words = set(doc)

    for w in fw:
        if w in words:
            d[w] = True
        else:
            d[w] = False

    return d


feature_sets = []

for doc, label in docs:
    feature_sets.append((features(doc), label))


train_data = feature_sets[:1500]
test_data = feature_sets[1500:]

print("Training samples:", len(train_data))
print("Testing samples:", len(test_data))
```

```python
# 25. Train a Naive Bayes classifier using the movie_reviews dataset
#and print the accuracy.
from nltk import NaiveBayesClassifier
from nltk.classify import accuracy

clf = NaiveBayesClassifier.train(train_data)

# step 7: calculate accuracy
acc = accuracy(clf, test_data)

print("Accuracy:", acc * 100, "%")
```

```python
# 26. Display the top 10 most informative features of the Naive Bayes classifier.
clf.show_most_informative_features(10)
```

```python
# 27. Write a function that takes a custom sentence and predicts if it is positive
#or negative using the trained model.

def predict_sentiment(sentence):
    words = word_tokenize(sentence.lower())
    feature = {}
    wset = set(words)
    for x in fw:                  
        feature[x] = (x in wset)
    return clf.classify(feature)
print(predict_sentiment("The film was boring and too long"))
```

```python
# 28. Save the trained Naive Bayes classifier using pickle 
#and load it again to make predictions.
import pickle

with open("naive_bayes_clf.pkl", "wb") as f:
    pickle.dump(clf, f)

with open("naive_bayes_clf.pkl", "rb") as f:
    loaded_clf = pickle.load(f)

sentence = "Everyday tones "


words = nltk.word_tokenize(sentence.lower())

feature = {}
wset = set(words)
for x in fw:
    feature[x] = (x in wset)

print("Prediction:", loaded_clf.classify(feature))
```

```python
# 30. Create a complete text classification pipeline using NLTK from raw 
#text to sentiment prediction


docs = []
for category in movie_reviews.categories():
    for fileid in movie_reviews.fileids(category):
        docs.append((list(movie_reviews.words(fileid)), category))

all_words = [w.lower() for w in movie_reviews.words()]
freq = FreqDist(all_words)
fw = list(freq.keys())[:2000]

def extract_features(words):
    words_set = set(words)
    features = {}
    for x in fw:
        features[x] = (x in words_set)
    return features

feature_sets = [(extract_features(w), c) for (w, c) in docs]
train_set, test_set = feature_sets[:1900], feature_sets[1900:]
clf = NaiveBayesClassifier.train(train_set)

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess(text):
    tokens = word_tokenize(text.lower())
    filtered = [w for w in tokens if w.isalpha() and w not in stop_words]
    lemmatized = [lemmatizer.lemmatize(w) for w in filtered]
    return lemmatized

def predict_sentiment(text):
    words = preprocess(text)
    features = extract_features(words)
    return clf.classify(features)

user_input = input("Enter a sentence to predict sentiment: ")
prediction = predict_sentiment(user_input)
print("Predicted sentiment:", prediction)
```
