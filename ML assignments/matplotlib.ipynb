```python
import matplotlib.pyplot as pp
import numpy as nm
```

```python
# Q1: Create a simple line plot of y = 2x + 3 for x values ranging from 0 to 10.
x=nm.linspace(0,10)
y=(2*x)+3

pp.plot(x,y)
pp.title('for y=2x+3')
pp.xlabel('X-axis')
pp.ylabel('Y-axis')
pp.show
```

```python
x=nm.linspace(0,10,100)
y1=nm.sin(x)
y2=nm.cos(x)

pp.plot(x,y1,color='red',linestyle='--',marker='o',label='sin(x)')
pp.plot(x,y2,color='blue',linestyle='-',marker='x',label='cos(x)')

pp.title('simple line plot of sin(x)')
pp.xlabel('X-axis')
pp.ylabel('Y-axis')


pp.grid(True)
pp.legend()

pp.show()
```

```python
# Q3: Create a bar chart showing the sales
#of four products: A, B, C, and D with random values.
x=['A','B','C','D']
y1=nm.random.rand(4)


pp.bar(x,y1,color='red')


pp.title('simple bar plot of ')
pp.xlabel('category')
pp.ylabel('value')


pp.show()
```

```python
# Q4: Generate a scatter plot with random x and y values
#between 0 and 1, and use different colors for points

x=nm.random.rand(50)
y=nm.random.rand(50)
col=nm.random.rand(50)
pp.scatter(x,y,c=col)
pp.title('sctter plot')
pp.xlabel('X')
pp.ylabel('y')
pp.show
```

```python
# Q5: Plot a histogram for 1000 random numbers generated from a normal distribution.
x=nm.random.randn(1000)

pp.hist(x,bins=30,color='purple',alpha=0.7,edgecolor='black')


pp.title('simple histogram plot  ')
pp.xlabel('X')
pp.ylabel('Y')


pp.show()
```

```python
# Q6: Draw a pie chart representing the percentage of five fruits: 
#Apple, Banana, Mango, Grapes, and Orange.
s=[25,15,30,20,10]
si=['apple','banana','mango','grapes','orange']
q=['red','yellow','orange','green','teal']
pp.pie(s,labels=si,colors=q,autopct='%1.1f%%',startangle=90,shadow=True)


pp.title('simple pie plot  ')
pp.show()
```

```python
# Q7: Create two subplots in one figure – one showing a sine wave and 
#the other showing a cosine wave.
x=nm.linspace(0,10,100)
y1=nm.sin(x)
y2=nm.cos(x)

pp.figure(figsize=(10,5))

pp.subplot(1,2,1)
pp.plot(x,y1,'r--')
pp.title('sin')

pp.subplot(1,2,2)
pp.plot(x,y2,'b-')
pp.title('cos')

pp.suptitle('multi')
pp.tight_layout()
pp.show
```

```python
# Q8: Customize a line plot by adding title, axis labels, grid lines, 
#and saving it as 'line_plot.png'.
x=nm.linspace(0,10,100)
y=nm.linspace(0,10,100)

pp.plot(x,y)
pp.title('simple line plot of sin(x)')
pp.xlabel('X-axis')
pp.ylabel('Y-axis')

pp.grid(True)

pp.savefig('line_plot.png',dpi=300,bbox_inches='tight')
pp.show()
```

```python
# Q9: Create a 2x2 grid of subplots showing different plots: line, bar, 
#scatter, and histogram.

fig,axs=pp.subplots(2,2,figsize=(10,8))

# line plot
x=nm.linspace(0,10,100)
axs[0,0].plot(x,nm.sin(x),'r-')
axs[0,0].set_title('sine')

#bar plot
ct=['a','b','c','d']
v=[10,15,7,12]
axs[0,1].bar(ct,v,color='teal')
axs[0,1].set_title('bar')

#hist
d=nm.random.randn(1000)
axs[1,0].hist(d,bins=25,color='purple',edgecolor='black')
axs[1,0].set_title('histogram')

#scatter
x=nm.random.rand(50)
y=nm.random.rand(50)
axs[1,1].scatter(x,y,color='red')
axs[1,1].set_title('scatter')


pp.suptitle('multi')
pp.tight_layout()
pp.savefig('xyz.png',dpi=300,bbox_inches='tight')
pp.show
```

```python
# Q10: Create a line plot for y = e^x for x between 0 and 5, 
#and change the color, marker, and line style.
x=nm.linspace(0,5,100)
y=nm.exp(x)

pp.plot(x,y,color='teal',marker='*',linestyle='--')

pp.title('simple exponential plot  ')
pp.xlabel('X')
pp.ylabel('x exp')

pp.savefig('abc.png',dpi=300,bbox_inches='tight')
pp.show
```
